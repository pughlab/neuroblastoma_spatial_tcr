# Step 6.5 Make sankey plot
This step plots a Sankey plot for the TCR plot filtration process.

```python
# Imports
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import matplotlib as mpl
import scanpy as sc
from pathlib import Path
```

```python
# Path config
figdir = '/path/to/integrated/figures/'
```

```python
# Figure settings
plt.rcParams['savefig.transparent'] = True
mpl.rcParams["font.family"] = "Helvetica"
mpl.rcParams['font.size'] = 6
sc.settings.figdir = figdir
```

```python
# Define palette and labels
palette = {
    "c1": "#f1eef6",
    "c2": "#d7b5d8",
    "c3": "#df65b0",
    "c4": "#dd1c77",
    "c5": "#980043",
}

labels = [
    "TCR probes<br>(100)",               # 0
    "Passed<br>(46)",                    # 1
    "Failed<br>criteria<br>(54)",        # 2
    "Second step<br>pass (100%)<br>(46)",# 3
    "Correct<br>chain<br>(24)",          # 4
    "Wrong<br>chain<br>(22)",            # 5
]
```

```python
# Helper function for plotting
def hex_to_rgba(h, a=0.5):
    h = h.lstrip("#")
    r, g, b = int(h[0:2], 16), int(h[2:4], 16), int(h[4:6], 16)
    return f"rgba({r},{g},{b},{a})"
```

```python
fig = go.Figure(
    data=[
        go.Sankey(
            arrangement="fixed",  # important: respect manual x/y
            node=dict(
                pad=70,
                thickness=10,
                line=dict(color="white", width=0),
                # label=labels,
                color=[
                    palette["c1"],  # 0 TCR probes
                    palette["c2"],  # 1 Passed
                    palette["c4"],  # 2 Failed criteria
                    palette["c2"],  # 3 Second-step pass
                    palette["c3"],  # 4 Correct chain
                    palette["c5"],  # 5 Wrong chain
                ],
                # columns: start -> split1 -> split2 -> final
                x=[0.02, 0.34, 0.34, 0.66, 0.98, 0.98],
                y=[0.50, 0.35, 0.70, 0.35, 0.25, 0.55],
            ),
            link=dict(
                source=[0, 0, 1, 3, 3],
                target=[1, 2, 3, 4, 5],
                value=[46, 54, 46, 24, 22],
                color=[
                    hex_to_rgba(palette["c2"], 0.5),
                    hex_to_rgba(palette["c4"], 0.5),
                    hex_to_rgba(palette["c2"], 0.5),
                    hex_to_rgba(palette["c3"], 0.5),
                    hex_to_rgba(palette["c5"], 0.5),
                ],
            ),
        )
    ]
)

fig.update_layout(font=dict(family="Helvetica", size=6, color="black"))

out_path = Path(figdir) / "sankey_tcr_probes.pdf"

fig.write_image(
    str(out_path),
    format="pdf",
    width=int(18 / 2.54 * 96),
    height=int(9.5 / 2.54 * 96),
    scale=1,
) 

fig.show()
```
