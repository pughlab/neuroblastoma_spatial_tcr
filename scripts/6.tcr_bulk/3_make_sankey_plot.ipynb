# Step 6.5 Make sankey plot
This step plots a Sankey plot for the TCR plot filtration process.

```python
# Imports
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import matplotlib as mpl
import scanpy as sc
from pathlib import Path
```

```python
# Path config
figdir = '/Users/yyj/Doc/1_dod_dec25/figures/'
```

```python
# Figure settings
plt.rcParams['savefig.transparent'] = True
mpl.rcParams["font.family"] = "Helvetica"
mpl.rcParams['font.size'] = 6
sc.settings.figdir = figdir
```

```python
# Define palette and labels
palette = {
    "c1": "#f1eef6",
    "c2": "#d7b5d8",
    "c3": "#df65b0",
    "c4": "#dd1c77",
    "c5": "#980043",
}

labels = [
    "TCR probes<br>(100)",
    "Passed<br>(46)",
    "Failed<br>criteria<br>(54)",
    "Correct<br>chain<br>(24)",
    "Wrong<br>chain<br>(22)",
]
```

```python
# Helper function for plotting
def hex_to_rgba(h, a=0.5):
    h = h.lstrip("#")
    r, g, b = int(h[0:2], 16), int(h[2:4], 16), int(h[4:6], 16)
    return f"rgba({r},{g},{b},{a})"
```

```python
fig = go.Figure(
    data=[
        go.Sankey(
            arrangement="snap",
            node=dict(
                pad=70,
                thickness=10,
                line=dict(color="white", width=0),
                #label=labels,
                color=[
                    palette["c1"],  # TCR probes
                    palette["c2"],  # Passed
                    palette["c4"],  # Failed
                    palette["c3"],  # Chain-matching
                    palette["c5"],  # Chain-nonmatching
                ],
            ),
            link=dict(
                source=[0, 0, 1, 1],
                target=[1, 2, 3, 4],
                value=[46, 54, 24, 22],
                color=[
                    hex_to_rgba(palette["c2"], 0.5),
                    hex_to_rgba(palette["c4"], 0.5),
                    hex_to_rgba(palette["c3"], 0.5),
                    hex_to_rgba(palette["c5"], 0.5),
                ],
            ),
        )
    ]
)

fig.update_layout(font=dict(family="Helvetica", size=6, color="black"))

out_path = Path(figdir) / "sankey_tcr_probes.pdf"



fig.write_image(str(out_path), format="pdf",
                width=int(18/2.54*96), height=int(9.5/2.54*96), scale=1)

fig.show()
```
